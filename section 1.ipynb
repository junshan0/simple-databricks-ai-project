```python
-- Central catalog for the whole video-AI project
CREATE CATALOG IF NOT EXISTS video_ai
  COMMENT 'Video processing RAG app - managed on Databricks Free Edition';

-- Comment: This creates a top-level catalog to group schemas (bronze/silver/gold/ai).
-- Run this in a SQL-capable notebook cell or a Databricks SQL editor.
```

```python
-- Medallion layers + a dedicated AI layer
CREATE SCHEMA IF NOT EXISTS video_ai.bronze COMMENT 'Raw ingested video metadata';
CREATE SCHEMA IF NOT EXISTS video_ai.silver COMMENT 'Chunks, captions, metadata';
CREATE SCHEMA IF NOT EXISTS video_ai.gold   COMMENT 'Curated, query-ready assets';
CREATE SCHEMA IF NOT EXISTS video_ai.ai     COMMENT 'Embeddings, search-prep, indexes';

-- Comment: Schemas organize data by processing stage. Use 'bronze' for raw ingests,
-- 'silver' for cleaned/chunked data, 'gold' for curated tables, and 'ai' for model artifacts.
```

```python
CREATE VOLUME IF NOT EXISTS video_ai.bronze.raw_videos
  COMMENT 'Landing zone for raw video files';

-- Comment: A volume is used as a storage landing zone. Replace with your workspace volume path
-- when running in Databricks.
```

```python
# Set local path for the volume (Databricks may use a different mount path)
volume_path = "/Volumes/video_ai/bronze/raw_videos"
# Create a training directory inside the volume if it does not exist
dbutils.fs.mkdirs(f"{volume_path}/training")
# List the contents to verify the directory was created
display(dbutils.fs.ls(volume_path))
```
