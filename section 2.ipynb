# Section 2: Setting Up an S3 Bucket for Videos

*Notes:* This section shows how to register an S3 bucket with Databricks as an external location. Ensure your AWS IAM role or credentials have the correct permissions before you begin.

The detailed background of this code is in this blog:

https://medium.com/@junshan0/setting-up-an-s3-bucket-for-videos-4874a270be3c?sk=cc76de8f6e6cc7c3178fe97efdce620a

*Note:* The blog includes guidance for creating an S3 bucket, configuring IAM policies, and setting up storage credentials.

```python
CREATE EXTERNAL LOCATION video_ai_workshop
  URL 's3://<bucket>/'
  WITH (STORAGE CREDENTIAL `video_ai_workshop_cred`)
  COMMENT 'Root location for workshop video files';

-- Comment: Replace `<bucket>` with your actual S3 bucket name and ensure the storage credential
-- `video_ai_workshop_cred` exists (created via Databricks or Unity Catalog). Run this in a SQL cell.
```

```python
-- List the files Databricks can now see
LIST 's3://<bucket>/';

-- Comment: Use this SQL command to verify that Databricks can access the external location.
-- You may need to run this in a SQL notebook cell depending on your Databricks runtime.
```

```python
# Use dbutils to list files programmatically (Python cell)
files = dbutils.fs.ls("s3://<bucket>/")
display(files)

# Comment: Replace `<bucket>` and ensure credentials/mounts are configured. Use this when you prefer Python over SQL.
```
